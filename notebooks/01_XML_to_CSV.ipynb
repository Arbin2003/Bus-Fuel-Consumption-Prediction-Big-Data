```python
import os
import xml.etree.ElementTree as ET
import pandas as pd

xml_folder = r"C:\Users\ASUS\Documents\Bus  project\data\datatransxchange_xml"

namespace = {"tx": "http://www.transxchange.org.uk/"}

rows = []
```

```python
xml_files = [f for f in os.listdir(xml_folder) if f.endswith(".xml")]

for file in xml_files:

    file_path = os.path.join(xml_folder, file)

    try:
        tree = ET.parse(file_path)
        root = tree.getroot()

        operator = root.find(".//tx:Operator/tx:TradingName", namespace)
        service = root.find(".//tx:ServiceCode", namespace)
        line = root.find(".//tx:LineName", namespace)
        departure = root.find(".//tx:DepartureTime", namespace)
        distance = root.find(".//tx:Distance", namespace)
        runtime = root.find(".//tx:RunTime", namespace)

        stop_count = len(root.findall(".//tx:StopPointRef", namespace))

        rows.append({
            "File_Name": file,
            "Operator": operator.text if operator is not None else None,
            "Service_Code": service.text if service is not None else None,
            "Line_Name": line.text if line is not None else None,
            "Departure_Time": departure.text if departure is not None else None,
            "Distance": distance.text if distance is not None else None,
            "Run_Time": runtime.text if runtime is not None else None,
            "Number_of_Stops": stop_count
        })

    except Exception as e:
        print("Error:", file, e)
```

```python
print("Total extracted rows:", len(rows))
```

```python
df = pd.DataFrame(rows)

print(df.shape)

df.head()
```
